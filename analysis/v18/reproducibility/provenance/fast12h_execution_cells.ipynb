## Twelve-hour availability sensitivity: cached observations and efficient summaries

The 12-hour sensitivity reuses checksummed, cleaned individual lab observations.
It does not try to reconstruct measurements from already-binned CNN arrays.
MIMIC availability is recalculated as recorded time plus 12 hours; eICU uses the
later of that minimum lag and the preserved revision availability. This increasing-lag
conversion is not used to infer an earlier release time.

The builder preserves the same history rules, one-hour draw clusters, revision-conflict
exclusions, reference intervals, discharge eligibility and three-hour raw tensors.
Shared empty-analyte templates, index-only lookups and observed-analyte calculations
avoid repeated empty-dataframe statistics. Duplicate-time statistics retain their
original aggregation. No labels are used by this optimization.

Real-source replay and edge fixtures check values, missingness, column order, dtypes,
raw tensors and eligibility counts against the reference builder. The execution
implementation and source checksums have independent checkpoint provenance; the
scientific study settings and existing model keys are unchanged. Each 250-admission
batch is saved atomically, so interrupted construction resumes completed batches.

This notebook uses the same output folder and retains completed primary, CNN and
sensitivity work. Stop the other kernel before running this notebook. Do not delete
checkpoints or change RUN_NAME. The optimization takes effect when this notebook's
definition cells run; editing a file does not replace functions in a running kernel.


```python
"""Equivalent, separately checkpointed 12-hour lab feature execution."""

V18_FAST12_SPEC = json.loads('{"checkpoints": {"admissions": {"path": "BE_integrated_grid_v15_outputs/lag_8h/feature_build_checkpoints/98d4ebfc6f9c385fb4bb50b8/admissions.pkl.gz", "sha256": "c9ffd238d25a4ae6ffdbf689bc897113b12771044bb8bdc32d67a1193ce8c719"}, "extracted_labs": {"path": "BE_integrated_grid_v15_outputs/lag_8h/feature_build_checkpoints/98d4ebfc6f9c385fb4bb50b8/extracted_labs.pkl.gz", "sha256": "77d3f877318e38877ce3f85d54f1f27d1e1090453f003d224f52a93e0c3d4302"}}, "feature_config": {"cnn_bin_hours": 3.0, "cnn_window_hours": 54.0, "first_draw_max_hours": 24.0, "history_slope_span_hours": 6.0, "min_core_analytes": 2, "min_slope_span_hours": 1.0, "release_lag_hours": 8.0, "repeat_min_gap_hours": 6.0, "second_draw_max_hours": 48.0, "set_width_hours": 1.0}, "old_lag": 8.0, "source_files": [{"bytes": 12548562, "modified_ns": 1775687457703726500, "path": "C:\\\\Users\\\\local-researcher\\\\Downloads\\\\Bacterial Endocarditis Prediction Pipeline\\\\ADMISSIONS.csv"}, {"bytes": 2628900, "modified_ns": 1775687468251007100, "path": "C:\\\\Users\\\\local-researcher\\\\Downloads\\\\Bacterial Endocarditis Prediction Pipeline\\\\PATIENTS.csv"}, {"bytes": 19137527, "modified_ns": 1775687457957057700, "path": "C:\\\\Users\\\\local-researcher\\\\Downloads\\\\Bacterial Endocarditis Prediction Pipeline\\\\DIAGNOSES_ICD.csv"}, {"bytes": 1854245647, "modified_ns": 1775687467973386900, "path": "C:\\\\Users\\\\local-researcher\\\\Downloads\\\\Bacterial Endocarditis Prediction Pipeline\\\\LABEVENTS.csv"}, {"bytes": 48277161, "modified_ns": 1525878609000000000, "path": "C:\\\\Users\\\\local-researcher\\\\Downloads\\\\Bacterial Endocarditis Prediction Pipeline\\\\eicu\\\\patients.csv"}, {"bytes": 283370401, "modified_ns": 1525817695000000000, "path": "C:\\\\Users\\\\local-researcher\\\\Downloads\\\\Bacterial Endocarditis Prediction Pipeline\\\\eicu\\\\diagnoses.csv"}, {"bytes": 2364239214, "modified_ns": 1525817908000000000, "path": "C:\\\\Users\\\\local-researcher\\\\Downloads\\\\Bacterial Endocarditis Prediction Pipeline\\\\eicu\\\\lab.csv"}, {"bytes": 664056, "modified_ns": 1778939647465300800, "path": "C:\\\\Users\\\\local-researcher\\\\Downloads\\\\Bacterial Endocarditis Prediction Pipeline\\\\CMS32_DESC_LONG_SHORT_DX.xlsx", "sha256": "e79455f7860e506e69df66e16bf0268d4a5a299bf5ef294ed0de628910c7c06e"}], "validation": {"additional_fixture_report_sha256": "6caf3b14c6e58db3492f788a20b50236414cf7f76a86a2016b5f210230ccb34d", "evidence": {"cohorts": {"MIMIC": {"accelerated_seconds": 9.3847481000048, "admissions": 43, "columns": 1048, "eligibility_flow_exact": true, "frame_dtypes_exact": true, "frame_values_exact": true, "lab_records": 7268, "patients": 22, "reference_seconds": 18.35116719999496, "rows": 78, "saved_observations_match_independent_raw_extraction": true, "speedup": 1.9554245893915445, "tensors_exact": true}, "eICU": {"accelerated_seconds": 5.485801199996786, "admissions": 58, "columns": 1048, "eligibility_flow_exact": true, "frame_dtypes_exact": true, "frame_values_exact": true, "lab_records": 3322, "patients": 30, "reference_seconds": 11.451516000000993, "rows": 95, "saved_observations_match_independent_raw_extraction": true, "speedup": 2.0874828639447784, "tensors_exact": true}}, "fixtures": {"MIMIC": {"accelerated_seconds": 1.5974123000050895, "admissions": 32, "columns": 1048, "eligibility_flow_exact": true, "frame_dtypes_exact": true, "frame_values_exact": true, "lab_records": 328, "outpatient_results_cross_preindex_availability_boundary": true, "patients": 8, "reference_seconds": 5.4503312000015285, "rows": 62, "speedup": 3.4119752301795616, "tensors_exact": true}, "decreasing_lag_rejected": true, "eICU": {"accelerated_seconds": 1.215705299997353, "admissions": 32, "columns": 1048, "eligibility_flow_exact": true, "frame_dtypes_exact": true, "frame_values_exact": true, "lab_records": 320, "patients": 8, "reference_seconds": 3.873217799999111, "rows": 62, "speedup": 3.1859841361286687, "tensors_exact": true}, "revision_adjustment": true, "shutdown_resume": {"completed_shard_preserved": true, "recomputed_rows": 50, "resumed_values_exact": true, "total_source_rows": 300}, "statistics": {"cases": 40, "duplicates_and_zero_span": true, "exact": true}}}, "integration": {"actual_jupyter_compilation": true, "changed_feature_settings_rejected": true, "changed_source_rejected": true, "definition_cell_reexecution_idempotent": true, "live_study_owner_recipe_unchanged": true, "only_12h_preparation_hooked": true, "primary_CNN_receipt_recipes_unchanged": true, "source_checksums_valid": true, "study_output_writes": 0}, "integration_report_sha256": "d1d67146694252d3c8c882204cbe3869051195659a8230a8d328c8d9f3580e00", "report_sha256": "e8c933949ed9bc877af6c99a8ab5fb6111231a3fe24e2d0b463a33cd0019dcfa", "scope": "Bounded real-source replay plus edge fixtures; not a full-cohort rebuild", "storage": {"both_cohorts": true, "canonical_cache_published": true, "complete_feature_preparation": true, "completed_cache_loaded_without_building": true, "directory": "C:\\\\Users\\\\local-researcher\\\\AppData\\\\Local\\\\Temp\\\\be_v18_fast12_validation_s7n3j62u", "provenance_written": true, "study_output_writes": 0, "values_dtypes_tensors_audit_preserved": true}, "storage_report_sha256": "4ab9556a0a8c005782ecf7c6217f09352c83d077f1f0ab9014de5d719e6be1df"}}')
if 'V18_FAST12_PREPARE_PARENT' not in globals():
    V18_FAST12_PREPARE_PARENT = V18_REUSE_PREPARE_PARENT


def v18_fast12_stats(z, min_span):
    # The canonical series has one observation per hour. Keep pandas' exact
    # duplicate-hour aggregation when duplicates really exist.
    if z.empty:
        return series_stats(z, min_span)
    if z.hour.duplicated().any():
        z = z.groupby('hour', as_index=False).value.mean().sort_values('hour')
    else:
        z = z.sort_values('hour')
    x = z.value.to_numpy(float)
    t = z.hour.to_numpy(float)
    span = t[-1] - t[0]
    d = dict(mean=x.mean(), median=np.median(x), min=x.min(), max=x.max(),
        sd=x.std(), iqr=np.quantile(x, .75)-np.quantile(x, .25),
        first=x[0], last=x[-1], slope_per_day=np.nan, change=np.nan,
        span_hours=float(span), count=float(len(x)), slope_valid=0.)
    if len(x) >= 2 and span >= min_span:
        centered = t-t.mean()
        d['slope_per_day'] = float(np.dot(centered, x-x.mean())/np.dot(centered, centered)*24)
        d['change'] = float(x[-1]-x[0])
        d['slope_valid'] = 1.
    return d


def v18_fast12_resolve(z, keys):
    # Unique keys need no groupby/copy. Ambiguous revisions use the unchanged
    # reference implementation, including exclusion of conflicting latest values.
    return resolve_revisions(z, keys) if z.duplicated(keys).any() else z


def v18_fast12_draws(z, cfg):
    if z.empty:
        return []
    core = z[z.analyte.isin(CORE)]
    times = np.sort(core.hour.dropna().unique())
    h = core.hour.to_numpy(float)
    available = core.release_hour.to_numpy(float)
    analytes = core.analyte.to_numpy()
    # Revisions are rare enough to check only the candidate cluster when needed.
    ambiguous = core.duplicated(['analyte', 'hour']).any()
    sets = []
    i = 0
    while i < len(times):
        t = float(times[i])
        limit = cfg['second_draw_max_hours'] if sets else cfg['first_draw_max_hours']
        if t > limit:
            break
        end = t+cfg['set_width_hours']
        keep = (h >= t) & (h < end) & (available <= end+cfg['release_lag_hours'])
        if ambiguous:
            g = v18_fast12_resolve(core.iloc[np.flatnonzero(keep)], ['analyte', 'hour'])
            count = g.analyte.nunique()
        else:
            count = len(np.unique(analytes[keep]))
        if count >= cfg['min_core_analytes']:
            if not sets:
                sets.append((t, end))
            elif t >= sets[0][0]+cfg['repeat_min_gap_hours']:
                sets.append((t, end))
                break
        i = int(np.searchsorted(times, end, side='left'))
    return sets


def v18_fast12_schema():
    stats = series_stats(pd.DataFrame(), 1.)
    history = {k:np.nan for k in ('mean','median','min','max','iqr','stay_mean',
        'within_stay_slope_mean','within_stay_slope_median')}
    history.update(count=0., stays=0.)
    base = {}
    for a in LABS:
        base.update({f'history__{a}__{k}':v for k,v in history.items()})
    base.update(age=np.nan, age_missing=1., sex_male=0., sex_unknown=1.,
        age_topcoded=0., current__elapsed_hours=np.nan, history__available=0.)
    for a in LABS:
        base.update({f'current__{a}__{k}':v for k,v in stats.items()})
        bf = boundary_features(np.nan, np.nan, np.nan, np.nan, np.nan, a in ('crp','esr'))
        base.update({f'current__{a}__{k}':v for k,v in bf.items()})
        base[f'combined__{a}__delta_history_median'] = np.nan
    names = list(base)
    positions = {name:i for i,name in enumerate(names)}
    current = {a:np.array([positions[f'current__{a}__{k}'] for k in stats]) for a in LABS}
    return names, positions, current, np.array(list(base.values()), dtype=float)


def v18_fast12_context(admissions, long):
    # Index arrays avoid holding two extra dictionaries of copied full lab frames.
    return dict(long=long, admissions=admissions,
        current=long.groupby('hospital_id', sort=False).indices,
        patients=long.groupby('patient_id', sort=False).indices,
        stays=admissions.groupby('patient_id', sort=False).indices,
        schema=v18_fast12_schema(), templates={})


def v18_fast12_batch(admissions, long, cohort, cfg, context):
    names, pos, current_pos, blank = context['schema']
    n_bins = int(np.ceil(cfg['cnn_window_hours']/cfg['cnn_bin_hours']))
    lab_names = list(LABS)
    lab_indices = {a:j for j,a in enumerate(lab_names)}
    rows=[]; metadata=[]; matrices=[]; valids=[]; event_times=[]; cutoffs=[]
    flow=Counter()
    for r in admissions.itertuples(index=False):
        flow['source_admissions'] += 1
        if not r.eligible:
            flow['ineligible_admissions'] += 1
            continue
        ids=context['current'].get(int(r.hospital_id))
        if ids is None or not len(ids):
            flow['no_current_labs'] += 1
            continue
        z=long.iloc[ids].copy()
        z['hour']=(z.time-r.start).dt.total_seconds()/3600 if cohort=='MIMIC' else z.time
        z['release_hour']=(z.available-r.start).dt.total_seconds()/3600 if cohort=='MIMIC' else z.available
        duration=(r.end-r.start).total_seconds()/3600 if cohort=='MIMIC' else r.duration
        z=z[z.hour.ge(0)&z.hour.le(duration)]
        sets=v18_fast12_draws(z,cfg)
        if not sets:
            flow['no_qualifying_first_set'] += 1
            continue
        ph=long.iloc[context['patients'][str(r.patient_id)]]
        earlier=context['admissions'].iloc[context['stays'][str(r.patient_id)]]
        if cohort=='MIMIC':
            earlier_ids=earlier.loc[earlier.end.lt(r.start)&earlier.start.lt(r.start)&earlier.valid_time,'hospital_id']
            hist=ph[(ph.hospital_id.isin(earlier_ids)|ph.hospital_id.isna())&ph.available.lt(r.start)&ph.time.lt(r.start)].copy()
            hist['hour']=(hist.time-pd.Timestamp('1970-01-01')).dt.total_seconds()/3600
            hist['episode']=hist.hospital_id.astype('string').fillna('outpatient_'+hist.time.dt.strftime('%Y-%m-%d'))
        else:
            earlier_ids=earlier.loc[earlier.year.lt(r.year)&earlier.valid_time,'hospital_id']
            hist=ph[ph.hospital_id.isin(earlier_ids)].copy()
            ends=earlier.set_index('hospital_id').duration
            hist=hist[hist.available.le(hist.hospital_id.map(ends))].copy()
            hist=v18_fast12_resolve(hist,['hospital_id','analyte','time'])
            hist['hour']=hist.time
            hist['episode']=hist.hospital_id.astype(str)
        key=((r.age>=18,r.age==18,r.age<19,r.age<50,r.age<60,r.age<=90)
            if np.isfinite(r.age) else None, str(r.sex), bool(r.age_topcoded))
        if key not in context['templates']:
            template=blank.copy()
            template[pos['age']]=r.age
            template[pos['age_missing']]=float(not np.isfinite(r.age))
            template[pos['sex_male']]=float(r.sex=='M')
            template[pos['sex_unknown']]=float(r.sex not in ('M','F'))
            template[pos['age_topcoded']]=float(r.age_topcoded)
            for a in lab_names:
                low, high=reference_interval(a,r.age,r.sex,r.age_topcoded)
                template[pos[f'current__{a}__reference_known']]=float(np.isfinite(low) and np.isfinite(high) and high>low)
            context['templates'][key]=template
        shared=context['templates'][key].copy()
        shared[pos['age']]=r.age
        shared[pos['history__available']]=float(len(hist)>0)
        # Empty-analyte statistics and validity flags come from one shared
        # template. Only actually observed analytes require statistical work.
        for a,az in hist.groupby('analyte'):
            if a not in lab_indices:
                continue
            values=az.value.to_numpy(float)
            h=dict(mean=values.mean(),median=np.median(values),min=values.min(),max=values.max(),
                iqr=np.quantile(values,.75)-np.quantile(values,.25),
                count=float(len(values)),stays=float(az.episode.nunique()))
            summaries=[v18_fast12_stats(g,cfg['history_slope_span_hours']) for _,g in az.groupby('episode')]
            h['stay_mean']=float(np.mean([s['mean'] for s in summaries]))
            slopes=[s['slope_per_day'] for s in summaries if s['slope_valid']]
            if slopes:
                h['within_stay_slope_mean']=float(np.mean(slopes))
                h['within_stay_slope_median']=float(np.median(slopes))
            for name,value in h.items():
                shared[pos[f'history__{a}__{name}']]=value
        for landmark,(draw_start,draw_end) in enumerate(sets,1):
            score_time=draw_end+cfg['release_lag_hours']
            if score_time>duration:
                flow[f'landmark_{landmark}_after_discharge'] += 1
                continue
            visible=z[z.hour.lt(draw_end)&z.release_hour.le(score_time)]
            visible=v18_fast12_resolve(visible,['hospital_id','analyte','hour'])
            features=shared.copy()
            features[pos['current__elapsed_hours']]=score_time
            seq=np.full((n_bins,len(lab_names)),np.nan,np.float32)
            seq_time=np.full_like(seq,np.nan)
            valid=(np.arange(n_bins)*cfg['cnn_bin_hours']<draw_end).astype(np.float32)
            for a,az in visible.groupby('analyte'):
                if a not in lab_indices:
                    continue
                s=v18_fast12_stats(az,cfg['min_slope_span_hours'])
                features[current_pos[a]]=list(s.values())
                low,high=reference_interval(a,r.age,r.sex,r.age_topcoded)
                bf=boundary_features(s['first'],s['last'],s['span_hours'] if s['slope_valid'] else np.nan,
                    low,high,a in ('crp','esr'))
                for name,value in bf.items():
                    features[pos[f'current__{a}__{name}']]=value
                features[pos[f'combined__{a}__delta_history_median']]=s['last']-shared[pos[f'history__{a}__median']]
                j=lab_indices[a]
                for q in az.sort_values('hour').itertuples():
                    b=int(q.hour//cfg['cnn_bin_hours'])
                    if 0<=b<n_bins:
                        seq[b,j]=q.value
                        seq_time[b,j]=q.hour
            rows.append(features.astype(np.float32))
            metadata.append(dict(patient_id=str(r.patient_id),hospital_id=int(r.hospital_id),
                label=int(r.label),landmark=landmark,score_hour=score_time))
            matrices.append(seq);valids.append(valid);event_times.append(seq_time);cutoffs.append(draw_end)
            flow[f'landmark_{landmark}_rows'] += 1
    if not rows:
        return None,None,dict(flow)
    frame=pd.concat([pd.DataFrame(metadata),pd.DataFrame(np.stack(rows),columns=names)],axis=1)
    seq=dict(values=np.stack(matrices),valid=np.stack(valids),hours=np.stack(event_times),cutoff=np.array(cutoffs))
    return frame,seq,dict(flow)


def v18_fast12_source(cfg):
    spec=V18_FAST12_SPEC
    if float(cfg.get('release_lag_hours',-1)) != 12.:
        return None
    if not V18_REUSE_READY:
        raise RuntimeError('12-hour acceleration requires the verified recovery preflight.')
    expected=spec['feature_config']
    for key in V18_REUSE_CATALOG['feature_keys']:
        if key!='release_lag_hours' and cfg[key]!=expected[key]:
            raise RuntimeError('12-hour acceleration input setting differs: '+key)
    if v18_recovery_signature()!=V18_REUSE_CATALOG['signature']:
        raise RuntimeError('12-hour acceleration source/preprocessing signature differs.')
    if input_manifest(DATA_ROOT,CFG)[0]['files'] != spec['source_files']:
        raise RuntimeError('12-hour acceleration source CSV/dictionary identities changed.')
    paths={name:v18_check_source(item) for name,item in spec['checkpoints'].items()}
    return paths


def v18_fast12_adjust(long,cohort,old_lag,new_lag):
    if new_lag<old_lag:
        raise ValueError('An increasing-lag cache cannot recover an earlier unrecorded revision time.')
    z=long.copy(deep=False)
    if cohort=='MIMIC':
        if not long.available.eq(long.time+pd.to_timedelta(old_lag,unit='h')).all():
            raise ValueError('MIMIC cleaned-lab availability does not match its pinned lag.')
        z['available']=long.time+pd.to_timedelta(new_lag,unit='h')
    else:
        if not (np.isfinite(long.available)&long.available.ge(long.time+old_lag)).all():
            raise ValueError('eICU cleaned-lab availability does not match its pinned minimum lag.')
        # max(t+12, max(t+8, revision)) == max(t+12, revision).
        z['available']=np.maximum(long.time+new_lag,long.available)
    return z


def v18_fast12_build(admissions,long,cohort,cfg,folder):
    folder=Path(folder)/'batch_250'
    folder.mkdir(parents=True,exist_ok=True)
    context=None;parts=[];sequences=[];flow=Counter()
    for start in range(0,len(admissions),250):
        stop=min(start+250,len(admissions))
        path=folder/f'{start:09d}_{stop:09d}.pkl'
        if path.is_file():
            saved=_v9_read_pickle(path)
            if saved['range']!=(start,stop) or saved['cohort']!=cohort:
                raise ValueError('Optimized batch metadata mismatch')
            frame,seq,counts=saved['result'];action='resumed'
        else:
            if context is None:
                context=v18_fast12_context(admissions,long)
            frame,seq,counts=v18_fast12_batch(admissions.iloc[start:stop],long,cohort,cfg,context)
            _v9_pickle(path,dict(range=(start,stop),cohort=cohort,result=(frame,seq,counts)))
            action='saved'
        flow.update(counts)
        if frame is not None:
            parts.append(frame);sequences.append(seq)
        print(f'[12h fast features {cohort}] {stop:,}/{len(admissions):,} admissions ({action})',flush=True)
    if not parts:
        raise ValueError(cohort+': no eligible 12-hour landmark rows')
    return pd.concat(parts,ignore_index=True),{key:np.concatenate([s[key] for s in sequences]) for key in sequences[0]},dict(flow)


def v18_fast12_prepare(root,out,cfg):
    if float(cfg.get('release_lag_hours',-1))!=12.:
        return V18_FAST12_PREPARE_PARENT(root,out,cfg)
    v18_assert_initialized();out=v18_owned_path(out)
    with v17_lock():
        paths=v18_fast12_source(cfg)
        manifest,fingerprint=v9_input_manifest(root,cfg)
        cache=out/'feature_cache'/fingerprint[:16]
        restored=_v9_load_final(cache,fingerprint)
        helpers=(v18_fast12_stats,v18_fast12_resolve,v18_fast12_draws,v18_fast12_schema,
            v18_fast12_context,v18_fast12_batch,v18_fast12_source,v18_fast12_adjust,v18_fast12_build,v18_fast12_prepare)
        execution=hashlib.sha256((fingerprint+v17_json(V18_FAST12_SPEC)+
            ''.join(v18_reuse_code(f) for f in helpers)).encode()).hexdigest()[:24]
        stage=out/'fast12_feature_checkpoints'/execution
        stage.mkdir(parents=True,exist_ok=True)
        provenance=dict(implementation='cutoff-aware observed-analyte execution',
            scientific_feature_fingerprint=fingerprint,execution_fingerprint=execution,
            source_checkpoints=V18_FAST12_SPEC['checkpoints'],
            validation=V18_FAST12_SPEC['validation'],
            changes='No cohort, cutoff, unit, feature, revision, model, fold or threshold definitions changed.')
        _v9_atomic(stage/'execution_manifest.json',lambda p:p.write_text(json.dumps(provenance,indent=2),encoding='utf-8'))
        if restored is None:
            done={cohort:stage/(cohort+'_features.pkl.gz') for cohort in ('MIMIC','eICU')}
            meta=stage/'audit_base.pkl.gz'
            if all(p.is_file() for p in done.values()) and meta.is_file():
                m,ms,mflow=_v9_read_pickle(done['MIMIC']);e,es,eflow=_v9_read_pickle(done['eICU'])
                source,unit_audit=_v9_read_pickle(meta)
            else:
                print('[12h fast features] Loading verified cleaned observations; raw CSV extraction is skipped.',flush=True)
                ma,ea,ep=_v9_read_pickle(paths['admissions']);del ep
                ml,el,unit_audit=_v9_read_pickle(paths['extracted_labs'])
                source=[dict(cohort=c,source_admissions=len(d),eligible_admissions=int(d.eligible.sum()),
                    eligible_cases=int(d.loc[d.eligible,'label'].sum()),eligible_patients=d.loc[d.eligible,'patient_id'].nunique())
                    for c,d in [('MIMIC',ma),('eICU',ea)]]
                _v9_pickle(meta,(source,unit_audit))
                if done['MIMIC'].is_file():
                    m,ms,mflow=_v9_read_pickle(done['MIMIC'])
                else:
                    ml=v18_fast12_adjust(ml,'MIMIC',V18_FAST12_SPEC['old_lag'],12.)
                    m,ms,mflow=v18_fast12_build(ma,ml,'MIMIC',cfg,stage/'MIMIC')
                    _v9_pickle(done['MIMIC'],(m,ms,mflow))
                del ml;gc.collect()
                if done['eICU'].is_file():
                    e,es,eflow=_v9_read_pickle(done['eICU'])
                else:
                    el=v18_fast12_adjust(el,'eICU',V18_FAST12_SPEC['old_lag'],12.)
                    e,es,eflow=v18_fast12_build(ea,el,'eICU',cfg,stage/'eICU')
                    _v9_pickle(done['eICU'],(e,es,eflow))
                del el,ma,ea;gc.collect()
            audit=dict(source=source,flow={'MIMIC':mflow,'eICU':eflow},units=unit_audit)
            _v9_publish_final(cache,fingerprint,manifest,m,ms,e,es,audit)
        else:
            m,ms,e,es,audit=restored
        cohort_summary(m,e).to_csv(out/'pre_age_filter_cohort_summary.csv',index=False)
        pd.DataFrame(audit['source']).to_csv(out/'source_admissions.csv',index=False)
        pd.DataFrame(audit['flow']).fillna(0).to_csv(out/'lab_eligibility_flow.csv')
        pd.DataFrame(audit['units']).fillna(0).to_csv(out/'unit_exclusion_counts.csv')
        # ICD construction is unchanged and remains independent of lab availability.
        build_icd_history(root,m,e,out)
        summary,_=analyzed_cohort_reports(m,e,cfg,out)
        reference_table().to_csv(out/'reference_intervals_with_sources.csv',index=False)
        key=str(out.resolve())
        V16_ACTIVE_FEATURE_MANIFEST[key]=manifest
        V16_ACTIVE_FEATURE_FINGERPRINT[key]=fingerprint
        _v9_atomic(out/'fast12_feature_provenance.json',lambda p:p.write_text(json.dumps(provenance,indent=2),encoding='utf-8'))
        v16_show_table('12-hour availability: analyzed cohort denominators',summary.round(4),V17_DESIGN_NOTE,max_rows=None)
        return m,ms,e,es,audit


# This execution-only delegate leaves every scientific function/owner hash and
# existing primary/CNN/8-hour model key intact. Its own implementation and source
# identities are bound independently in fast12_feature_checkpoints.
V18_REUSE_PREPARE_PARENT=v18_fast12_prepare
print('[12h fast features] Verified cleaned observations + resumable observed-analyte summaries; other analyses unchanged.',flush=True)
```
